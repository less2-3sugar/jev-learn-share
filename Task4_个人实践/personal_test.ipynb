```python

expected_results = [
    "自动分流：退款处理组",
    "转人工处理：置信度不足",
    "转人工处理：未知类别",
    "输入错误：置信度必须是数字",
    "输入错误：阈值必须是数字",
    "配置错误：阈值必须在0到1之间",
    "自动分流：退款处理组",
]

print("预期结果数量：", len(expected_results))
```

```python

test_cases = [
    ("refund", 0.90, 0.80),
    ("refund", 0.60, 0.80),
    ("logistics", 0.95, 0.80),
    ("refund", "0.85", 0.80),
    ("refund", 0.85, True),
    ("refund", 0.85, 0),
    ("refund", 0.80, 0.80),
]

print(test_cases)

ROUTES={
    "refund":"退款处理组",
    "complaint":"投诉处理组",
    "account_issue":"账户支持组",
    "order_status":"订单查询组"
}
def route_by_confidence(predicted, confidence, threshold):
    print("预测类别：", predicted)
    print("分类置信度：", confidence)
    print("分流阈值：", threshold)
    if isinstance(threshold, bool) or not isinstance(threshold, (int, float)):
        return "输入错误：阈值必须是数字"
    if threshold <= 0 or threshold > 1:
        return "配置错误：阈值必须在0到1之间"
    if isinstance(confidence, bool) or not isinstance(confidence, (int, float)):
        return "输入错误：置信度必须是数字"
    if confidence>1 or confidence<0:
        return "转人工处理：置信度数值不合法"
    if predicted not in ROUTES:
        return "转人工处理：未知类别"
    elif confidence>=threshold:
        return "自动分流："+ROUTES[predicted]
    else:
        return "转人工处理：置信度不足"
    
```

```python
print("测试案例数量：", len(test_cases))
print("预期结果数量：", len(expected_results))
```

```python
def run_tests(test_cases, expected_results):
    pass_count=0
    fail_cases=[]
    for case, expected in zip(test_cases, expected_results):
        predicted, confidence, threshold = case
        actual = route_by_confidence(predicted, confidence, threshold)
        try:
            assert actual == expected, f"预期={expected}，实际={actual}"
            pass_count += 1
        except AssertionError as error:
            fail_cases.append({
                "case": case,
                "expected": expected,
                "actual": actual,
                "error": str(error)
            })
    print("测试总数：", len(test_cases))
    print("通过数量：", pass_count)
    print("失败数量：", len(fail_cases))

    if test_cases:
        print(f"通过率：{pass_count / len(test_cases):.1%}")
    if fail_cases:
        print("失败案例详情：")
        for item in fail_cases:
            print("输入：", item["case"])
            print("预期：", item["expected"])
            print("实际：", item["actual"])
    return {
        "total": len(test_cases),
        "passed": pass_count,
        "failed": len(fail_cases),
        "fail_cases": fail_cases
    }
run_tests(test_cases, expected_results)
```

```python
def all_tests_passed(report):
    if not isinstance(report, dict):
        return False
    if "failed" not in report:
        return False
    failed_count = report["failed"]
    if not isinstance(failed_count, int) or isinstance(failed_count, bool):
        return False
    if failed_count < 0:
        return False
    return report["failed"] == 0
```

```python
report_test_cases = [
    ({"failed": 0}, True),
    ({"failed": 1}, False),
    ({"failed": -1}, False)
]
```

```python

pass_count = 0
fail_count = 0
fail_cases = []

test_cases_with_error = report_test_cases.copy()
test_cases_with_error[0] = ({"failed": 0}, False)

for test_input, expected in report_test_cases:
    # 1. 获取实际结果
    actual = all_tests_passed(test_input)

    # 2. 比较实际结果和预期结果
    passed = actual == expected

    print("测试输入：", test_input)
    print("预期结果：", expected)
    print("实际结果：", actual)
    print("是否通过：", passed)

    # 3. 统计通过和失败数量
    if passed:
        pass_count += 1
    else:
        fail_count += 1
        fail_cases.append({
            "input": test_input,
            "expected": expected,
            "actual": actual
        })

    print("-" * 30)

# 4. 输出测试统计报告
total_count = len(report_test_cases)

print("测试总数：", total_count)
print("通过数量：", pass_count)
print("失败数量：", fail_count)

# 5. 计算通过率，避免除以零
if total_count > 0:
    pass_rate = pass_count / total_count
    print(f"通过率：{pass_rate:.1%}")
else:
    print("通过率：暂无测试案例")

# 6. 输出失败案例详情
if fail_cases:
    print("\n失败案例详情：")
    for item in fail_cases:
        print("输入：", item["input"])
        print("预期：", item["expected"])
        print("实际：", item["actual"])
        print("-" * 30)
else:
    print("\n所有测试案例均通过！")
```
